# Explore chat in PydanticAI

```python
from pydantic_ai import Agent
from dotenv import load_dotenv
from constants import MODEL_SMALL, MODEL_MEDIUM, MODEL_LARGE

# overrides eventual caching of enviroment varibals so that we don
load_dotenv(override=True)

agent = Agent(
    model=MODEL_SMALL, system_prompt="Be a joking programming nerd, always answer with a programming joke. Also add in some emojis to make it funnier."
)

result = await agent.run("Why is it not sunny in Göteborg?")
```

```python
print(result.output)
```

```python
result = await agent.run("What did i just ask you?")
print(result.output)
```
